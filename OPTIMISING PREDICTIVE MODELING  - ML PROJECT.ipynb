```python
df_mean = df.copy()
df_knn = df.copy()
```

```python
#Fill Missing Values in df_mean

# Education (Mode)
df_mean['education'].fillna(df_mean['education'].mode()[0], inplace=True)

# Cigarettes Per Day (Median)
df_mean['cigsPerDay'].fillna(df_mean['cigsPerDay'].median(), inplace=True)

# Blood Pressure Medication (Mode)
df_mean['BPMeds'].fillna(df_mean['BPMeds'].mode()[0], inplace=True)

# Total Cholesterol (Mean)
df_mean['totChol'].fillna(df_mean['totChol'].mean(), inplace=True)

# BMI (Mean)
df_mean['BMI'].fillna(df_mean['BMI'].mean(), inplace=True)

# Heart Rate (Mean)
df_mean['heartRate'].fillna(df_mean['heartRate'].mean(), inplace=True)

# Glucose (Mean)
df_mean['glucose'].fillna(df_mean['glucose'].mean(), inplace=True)
```

```python
df_mean.isnull().sum()
```

```python
# filling missing values of df_knn - 
```

```python
from sklearn.impute import KNNImputer
imputer = KNNImputer(n_neighbors=5)
```

```python
df_knn = pd.DataFrame(
    imputer.fit_transform(df_knn),
    columns=df_knn.columns
)
```

```python
df_knn.isnull().sum()
```

```python
X_mean = df_mean.drop("TenYearCHD", axis=1)
y_mean = df_mean["TenYearCHD"]
```

```python
X_knn = df_knn.drop("TenYearCHD", axis=1)
y_knn = df_knn["TenYearCHD"]
```

```python
print("X_mean Shape:", X_mean.shape)
print("y_mean Shape:", y_mean.shape)

print("X_knn Shape:", X_knn.shape)
print("y_knn Shape:", y_knn.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
#Split df_mean
X_train_mean, X_test_mean, y_train_mean, y_test_mean = train_test_split(
    X_mean,
    y_mean,
    test_size=0.2,
    random_state=42,
    stratify=y_mean
)
```

```python
# split df_knn
X_train_knn, X_test_knn, y_train_knn, y_test_knn = train_test_split(
    X_knn,
    y_knn,
    test_size=0.2,
    random_state=42,
    stratify=y_knn
)
```

```python
print(X_train_mean.shape)
print(X_test_mean.shape)

print(y_train_mean.shape)
print(y_test_mean.shape)
```

```python
# not all machine elarning models require feature scaling hence we will just scale logistic regression
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
X_train_mean_scaled = scaler.fit_transform(X_train_mean)
X_test_mean_scaled = scaler.transform(X_test_mean)
```

```python
# Scale the KNN-Imputed Dataset

scaler_knn = StandardScaler()

X_train_knn_scaled = scaler_knn.fit_transform(X_train_knn)
X_test_knn_scaled = scaler_knn.transform(X_test_knn)
```

```python
from sklearn.linear_model import LogisticRegression

lr_mean = LogisticRegression(random_state=42, max_iter=1000)
```

```python
lr_mean.fit(X_train_mean_scaled, y_train_mean)
```

```python
y_pred_mean = lr_mean.predict(X_test_mean_scaled)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
print("Accuracy:", accuracy_score(y_test_mean, y_pred_mean))
```

```python
print("Precision:", precision_score(y_test_mean, y_pred_mean))
print("Recall:", recall_score(y_test_mean, y_pred_mean))
print("F1 Score:", f1_score(y_test_mean, y_pred_mean))
```

```python
cm = confusion_matrix(y_test_mean, y_pred_mean)
print(cm)
```

```python
print(classification_report(y_test_mean, y_pred_mean))
```

## Report Interpretation

You can write this in your report:

Logistic Regression achieved an overall accuracy of 84.79%. However, due to the imbalance in the target variable, accuracy alone is not sufficient to evaluate the model. The model produced a precision of 50%, indicating that half of the predicted positive cases were correct. However, the recall was only 6.98%, meaning that the model failed to identify most patients who developed coronary heart disease. Consequently, the F1-score was also low (12.24%). These results suggest that Logistic Regression, without additional techniques to address class imbalance, is not effective for identifying high-risk patients.

# Why are we doing this?

We'll use:

* Scaled SMOTE → Logistic Regression
* Unscaled SMOTE → Decision Tree & Random Forest

This is the standard practice because:

* Logistic Regression depends on feature scale.
* Decision Trees and Random Forests do not.

# Now Let's Create a Professional ROC Comparison Graph

Instead of having three separate ROC graphs, we'll create one graph comparing the three most important models:

* Logistic Regression + SMOTE
* Decision Tree + SMOTE
* Random Forest + SMOTE

```python
# Logistic Regression
y_prob_lr = lr_smote.predict_proba(X_test_mean_scaled)[:, 1]

# Decision Tree
y_prob_dt = dt_smote.predict_proba(X_test_mean)[:, 1]

# Random Forest
y_prob_rf = rf_smote.predict_proba(X_test_mean)[:, 1]
```

```python
from sklearn.metrics import roc_curve, roc_auc_score

# Logistic Regression
fpr_lr, tpr_lr, _ = roc_curve(y_test_mean, y_prob_lr)
auc_lr = roc_auc_score(y_test_mean, y_prob_lr)

# Decision Tree
fpr_dt, tpr_dt, _ = roc_curve(y_test_mean, y_prob_dt)
auc_dt = roc_auc_score(y_test_mean, y_prob_dt)

# Random Forest
fpr_rf, tpr_rf, _ = roc_curve(y_test_mean, y_prob_rf)
auc_rf = roc_auc_score(y_test_mean, y_prob_rf)
```

# Conclusion 



This project developed machine learning models to predict the risk of developing coronary heart disease within ten years using the Framingham Heart Study dataset. Data preprocessing included missing value imputation, exploratory data analysis, feature scaling, and handling class imbalance with SMOTE. Three machine learning algorithms—Logistic Regression, Decision Tree, and Random Forest—were trained and evaluated. Among all the models, Logistic Regression combined with SMOTE achieved the highest recall and ROC-AUC score, making it the most suitable model for identifying high-risk patients. The results demonstrate that addressing class imbalance significantly improves the detection of heart disease cases. This study highlights the importance of appropriate preprocessing, model evaluation, and the use of recall-focused metrics in healthcare prediction problems.

# Business/Medical Recommendation



The developed prediction model can assist healthcare professionals in identifying individuals at high risk of developing coronary heart disease. Early identification enables timely preventive measures such as lifestyle modification, regular monitoring, dietary changes, and medical intervention, potentially reducing the incidence of cardiovascular disease.
